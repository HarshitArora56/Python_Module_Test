```python
# Ques 1 Longest Increasing Subsequence

arr = [10,9,2,5,3,7,101,18]

# i
def longest_increasing_subsequence(arr):
    res = {}

    for i in range(len(arr)):
        l = []
        l.append(arr[i])
        for j in range(i+1, len(arr)):
            if arr[j] > l[-1]:
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = 1
                else:
                    res[len(l)] += 1
            elif len(l) > 2 and arr[j] > l[-2]:
                l.pop()
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = 1
                else:
                    res[len(l)] += 1
                

    return max(res.keys())


# ii
def count_lis(arr):
    res = {}

    for i in range(len(arr)):
        l = []
        l.append(arr[i])
        for j in range(i+1, len(arr)):
            if arr[j] > l[-1]:
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = 1
                else:
                    res[len(l)] += 1
            elif len(l) > 2 and arr[j] > l[-2]:
                l.pop()
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = 1
                else:
                    res[len(l)] += 1

    return res[max(res.keys())]

# iii
def get_lis_elements(arr):
    res = {}

    for i in range(len(arr)):
        l = []
        l.append(arr[i])
        for j in range(i+1, len(arr)):
            if arr[j] > l[-1]:
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = l
                else:
                    res[len(l)] = l

            elif len(l) > 2 and arr[j] > l[-2]:
                l.pop()
                l.append(arr[j])
                if len(l) not in res:
                    res[len(l)] = l
                else:
                    res[len(l)] = l


    return res[max(res.keys())]

                    
```

```python
print(f'LIS length is: {longest_increasing_subsequence(arr)}')
```

```python
print(f'Count LIS: {count_lis(arr)}')
```

```python
print(f'Get LIS Elements: {get_lis_elements(arr)}')
```

```python
# Ques 2 Fibonacci Modified Sequence

# i
def fibonacci_modified(t1,t2,n):
    if n == 0:
        return 'Error: No of terms cannot be zero'
    if n == 1:
        return 0
    if n == 2:
        return 1

    if n > 2:
        return fibonacci_modified(t1, t2, n-1) + fibonacci_modified(t1, t2, n-2) ** 2



# ii
def sum_range(t1,t2,start,end):
    sum_r = 0
    for i in range(start, end + 1):
        sum_r += fibonacci_modified(t1,t2,i)
    return sum_r


# iii
def count_divisible(t1,t2,n,k):
    count = 0
    for i in range(2,n+1): # not counting  0 based on sample output 
        if fibonacci_modified(t1,t2,i) % k == 0:
            count += 1
    return count
```

```python
t1 = int(input('t1 : '))
t2 = int(input('t2 : '))
n = int(input('n : '))
print(f'{n}th term of fibonacci_modified = {fibonacci_modified(t1,t2,n)}')
```

```python
start = int(input('Start : '))
end = int(input('End : '))
print(f'Sum_range from {start} to {end} = {sum_range(t1,t2,start,end)}')
```

```python
N = int(input('N : '))
k = int(input('k : '))
print(f'Count of terms divisible by {k} in first {N} terms (excluding 0) = {count_divisible(t1,t2,N,k)}')
```
